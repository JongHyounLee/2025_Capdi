```python
# ===== 3-Class TCN (drop label==2) — Cube.AI-friendly =====
# - 입력: (L=128, C=30) = 5 IMUs × (ax,ay,az,gx,gy,gz)
# - 전처리: 선형 리샘플 → 채널별 z-score(μ/σ 저장; label 2 제거 후 재계산)
# - 모델: Conv1D+BN+ReLU + Residual 1블록 → GAP1D → Dense(softmax, 3)
# - 출력: final_export_cubeai.h5, tcn_norm_stats.json(라벨 매핑 포함)

from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os, json
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ---------- 경로/설정 ----------
BASE = "/content/drive/MyDrive/imu_data"
CSV_PATHS = [
    f"{BASE}/imu_label_0_jonghyeon.csv",
    f"{BASE}/imu_label_1_jungro.csv",
    f"{BASE}/IMU_label_2_jeahyun.csv",   # 있어도 괜찮음: 아래에서 label==2 샘플 제거함
    f"{BASE}/imu_label_3_jonghyun.csv",
]
L = 128
NUM_CLASSES = 3  # ★ 3 클래스로 변경

# 채널 순서(고정)
axes = ["ax","ay","az","gx","gy","gz"]
CHANNELS = [f"IMU{k}_{a}" for k in range(1,6) for a in axes]  # 30개

# ---------- 유틸 ----------
def _pick_cols(df):
    # *_tick 제거, 대소문자 허용
    df = df[[c for c in df.columns if "_tick" not in c.lower()]]
    low = {c.lower(): c for c in df.columns}
    cols = []
    for c in CHANNELS:
        key = c.lower()
        if key not in low:
            raise ValueError(f"필수 컬럼 누락: {c}")
        cols.append(low[key])
    # move/label 표준화
    if "move" not in df.columns:
        m = next((low[k] for k in low if k=="move"), None)
        if m is None: raise ValueError("move 컬럼 필요")
        df.rename(columns={m:"move"}, inplace=True)
    if "label" not in df.columns:
        l = next((low[k] for k in low if k=="label"), None)
        if l is None: raise ValueError("label 컬럼 필요")
        df.rename(columns={l:"label"}, inplace=True)
    return df[cols + ["move","label"]]

def _resample_to_L(x_T_C, L=128):
    # x: [T,C] → 선형보간으로 [L,C]
    T, C = x_T_C.shape
    if T < 2:
        return np.repeat(x_T_C, L, axis=0)
    t_src = np.linspace(0.0, 1.0, num=T, dtype=np.float32)
    t_dst = np.linspace(0.0, 1.0, num=L, dtype=np.float32)
    out = np.empty((L, C), dtype=np.float32)
    for c in range(C):
        out[:, c] = np.interp(t_dst, t_src, x_T_C[:, c])
    return out

def build_dataset(paths):
    X_list, y_list = [], []
    for p in paths:
        if not os.path.exists(p):
            print(f"[WARN] not found: {p}")
            continue
        df = pd.read_csv(p)
        df = _pick_cols(df)
        for mv, g in df.groupby("move"):
            x = g[CHANNELS].astype(np.float32).values  # [T,30]
            x = _resample_to_L(x, L=L)                # [L,30]
            X_list.append(x)
            # 무브 내 라벨 혼재 시 다수결(평균 반올림)
            y_list.append(int(np.round(g["label"].astype(int).mean())))
    X = np.stack(X_list, axis=0)   # [N,L,30]
    y = np.array(y_list, dtype=np.int32)
    return X, y

# ---------- 데이터 로드 ----------
X, y = build_dataset(CSV_PATHS)
print("Original dataset:", X.shape, "class dist:", np.bincount(y))

# ---------- label==2 제거 & 라벨 재맵핑 ----------
DROP_LABEL = 2
mask = (y != DROP_LABEL)
X, y = X[mask], y[mask]

# 원하는 의미 체계에 맞게 매핑: 0->0, 1->1, 3->2
label_map = {0:0, 1:1, 3:2}
y = np.vectorize(label_map.get)(y)

print("Filtered dataset:", X.shape, "class dist (0,1,2):", np.bincount(y))

# ---------- 정규화(μ/σ) — 필터링 후 데이터로 재계산 ----------
flat = X.reshape(-1, X.shape[-1]).astype(np.float32)
mu = flat.mean(axis=0)
sigma = flat.std(axis=0)
sigma[sigma < 1e-6] = 1.0
X = (X - mu) / sigma


# μ/σ + 라벨매핑 저장(펌웨어에서 동일 적용)
payload = {
    "channels": CHANNELS,
    "mu": mu.tolist(),
    "sigma": sigma.tolist(),
    "label_map_old2new": {str(k): int(v) for k, v in label_map.items()},
    "dropped_label": int(DROP_LABEL),
}
with open(f"{BASE}/tcn_norm_stats.json", "w", encoding="utf-8") as f:
    json.dump(payload, f, ensure_ascii=False, indent=2)  # ← fp=f 추가
# ---------- split ----------
rng = np.random.RandomState(42)
idx = np.arange(len(X)); rng.shuffle(idx)
split = int(len(X)*0.8)
tr, va = idx[:split], idx[split:]
Xtr, ytr = X[tr], y[tr]
Xva, yva = X[va], y[va]

# ---------- 모델 ----------
def tcn_minimal(input_shape, num_classes=3, ch=32, k=5, p=0.1):
    inp = keras.Input(shape=input_shape, name="imu")      # [L,30]
    # stem
    x = layers.Conv1D(ch, k, padding="same", use_bias=False)(inp)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    if p>0: x = layers.Dropout(p)(x)
    # residual 1 block
    h = layers.Conv1D(ch, k, padding="same", use_bias=False)(x)
    h = layers.BatchNormalization()(h)
    h = layers.ReLU()(h)
    h = layers.Conv1D(ch, k, padding="same", use_bias=False)(h)
    h = layers.BatchNormalization()(h)
    x = layers.Add()([x, h])
    x = layers.ReLU()(x)
    # head
    x = layers.GlobalAveragePooling1D()(x)
    out = layers.Dense(num_classes, activation="softmax")(x)
    return keras.Model(inp, out, name="tcn_minimal")

model = tcn_minimal((L, 30), NUM_CLASSES)
model.compile(optimizer=keras.optimizers.Adam(1e-3),
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
model.summary()

# ---------- 학습 ----------
cb = [keras.callbacks.EarlyStopping(patience=8, restore_best_weights=True, monitor="val_accuracy")]
model.fit(Xtr, ytr, validation_data=(Xva, yva), epochs=40, batch_size=32, verbose=2, callbacks=cb)

# ---------- 내보내기 (Cube.AI 임포트) ----------
save_path = f"{BASE}/final_export_cubeai.h5"
model.save(save_path)
print("Saved:", save_path, "| Stats:", f"{BASE}/tcn_norm_stats.json")
```

```python
# === Generate C preprocessing files from stats JSON ===
import json, textwrap, os
BASE = "/content/drive/MyDrive/imu_data"   # 위 학습에 쓴 경로와 동일
L, C = 128, 30

with open(f"{BASE}/tcn_norm_stats.json","r",encoding="utf-8") as f:
    stats = json.load(f)
mu = stats["mu"]; sigma = stats["sigma"]

def fmt(arr):
    return ",\n  ".join([f"{float(x):.6f}f" for x in arr])

h = f"""#ifndef CUBEAI_PREPROCESS_H
#define CUBEAI_PREPROCESS_H
#include <stdint.h>
#ifdef __cplusplus
extern "C" {{
#endif
#define TCN_L {L}
#define TCN_C {C}
void tcn_linear_resample(const float* src, int src_len, int C, float* dst, int L);
void tcn_normalize_inplace(float* x, int L, int C);
const float* tcn_get_mu(void);
const float* tcn_get_sigma(void);
#ifdef __cplusplus
}}
#endif
#endif
"""

c = f"""#include "cubeai_preprocess.h"
#include <math.h>
#include <stddef.h>

static const float kMu[TCN_C] = {{
  {fmt(mu)}
}};

static const float kSigma[TCN_C] = {{
  {fmt(sigma)}
}};

// src: [src_len, C] row-major → dst: [L, C]
void tcn_linear_resample(const float* src, int src_len, int C, float* dst, int L) {{
    if (src_len <= 0) return;
    for (int i = 0; i < L; ++i) {{
        float pos = (src_len - 1) * (float)i / (float)(L - 1);
        int i0 = (int)pos;
        int i1 = i0 + 1; if (i1 >= src_len) i1 = src_len - 1;
        float t = pos - (float)i0;
        const float* s0 = src + (size_t)i0 * C;
        const float* s1 = src + (size_t)i1 * C;
        float* d = dst + (size_t)i * C;
        for (int c = 0; c < C; ++c) d[c] = s0[c]*(1.0f - t) + s1[c]*t;
    }}
}}

void tcn_normalize_inplace(float* x, int L, int C) {{
    const float eps = 1e-6f;
    for (int i = 0; i < L; ++i) {{
        float* row = x + (size_t)i * C;
        for (int c = 0; c < C; ++c) {{
            float s = kSigma[c]; if (s < eps) s = 1.0f;
            row[c] = (row[c] - kMu[c]) / s;
        }}
    }}
}}

const float* tcn_get_mu(void)    {{ return kMu;    }}
const float* tcn_get_sigma(void) {{ return kSigma; }}
"""

with open(f"{BASE}/cubeai_preprocess.h","w") as f: f.write(h)
with open(f"{BASE}/cubeai_preprocess.c","w") as f: f.write(c)

print("Created:")
print(f" - {BASE}/cubeai_preprocess.h")
print(f" - {BASE}/cubeai_preprocess.c")
```
